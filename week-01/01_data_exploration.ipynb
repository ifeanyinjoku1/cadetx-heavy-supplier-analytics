```python
import pandas as pd
import numpy as np

# Load core relational datasets
branches_df = pd.read_csv('/content/branches.csv')
customers_df = pd.read_csv('/content/customers.csv')
inventory_df = pd.read_csv('/content/inventory_master.csv')
invoices_df = pd.read_csv('/content/invoices.csv')
payments_df = pd.read_csv('/content/payments.csv')
products_df = pd.read_csv('/content/products.csv')
purchase_orders_header_df = pd.read_csv('/content/purchase_orders_header.csv')
purchase_orders_lines_df = pd.read_csv('/content/purchase_orders_lines.csv')
sales_orders_header_df = pd.read_csv('/content/sales_orders_header.csv')
sales_orders_lines_df = pd.read_csv('/content/sales_orders_lines.csv')
stock_ledger_df = pd.read_csv('/content/stock_ledger.csv')
suppliers_df = pd.read_csv('/content/suppliers.csv')
```

```python
# Dictionary to manage all datasets efficiently
datasets = {
    'branches': branches_df,
    'customers': customers_df,
    'inventory': inventory_df,
    'invoices': invoices_df,
    'payments': payments_df,
    'products': products_df,
    'purchase_headers': purchase_orders_header_df,
    'purchase_lines': purchase_orders_lines_df,
    'sales_headers': sales_orders_header_df,
    'sales_lines': sales_orders_lines_df,
    'stock_ledger': stock_ledger_df,
    'suppliers': suppliers_df
}
```

## **1. Dataset Overview & Missing Values**

```python
# High-Level Summary & Structural Health Check
print("=== DATASET OVERVIEW & MISSING VALUES ===")
overview_list = []

for name, df in datasets.items():
    overview_list.append({
        'Dataset': name,
        'Rows': df.shape[0],
        'Columns': df.shape[1],
        'Duplicate Rows': df.duplicated().sum(),
        'Total Missing Values': df.isnull().sum().sum(),
        'Memory Usage (MB)': round(df.memory_usage(deep=True).sum() / (1024 ** 2), 2)
    })

overview_df = pd.DataFrame(overview_list)
print(overview_df.to_string(index=False))
print("-" * 60)
```

# **2. Detailed Column Profiling & Null Audit Function**

```python

def profile_dataset(df_name):
    df = datasets[df_name]
    print(f"\n--- Profiling: {df_name.upper()} ---")

    summary = pd.DataFrame({
        'Data Type': df.dtypes,
        'Null Count': df.isnull().sum(),
        'Null %': (df.isnull().sum() / len(df) * 100).round(2),
        'Unique Values': df.nunique()
    })
    print(summary)

# Example: Profile core transactional tables
profile_dataset('sales_lines')
profile_dataset('inventory')
```

# **3. Numeric & Financial Summary Statistics**

```python
# 3. Numeric & Financial Summary Statistics
print("\n=== SALES & ORDER FINANCIAL SUMMARY ===")
if 'total_amount' in sales_orders_header_df.columns:
    print(sales_orders_header_df[['total_amount']].describe().round(2))
elif 'unit_price' in sales_orders_lines_df.columns and 'quantity' in sales_orders_lines_df.columns:
    sales_orders_lines_df['line_total'] = sales_orders_lines_df['unit_price'] * sales_orders_lines_df['quantity']
    print(sales_orders_lines_df[['quantity', 'unit_price', 'line_total']].describe().round(2))
```
