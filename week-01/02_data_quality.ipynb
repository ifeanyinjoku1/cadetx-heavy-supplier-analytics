```python
import pandas as pd
import numpy as np

# Load core relational datasets
branches_df = pd.read_csv('branches.csv')
inventory_df = pd.read_csv('inventory_master.csv')
suppliers_df = pd.read_csv('suppliers.csv')
products_df = pd.read_csv('products.csv')
sales_df = pd.read_csv('sales_orders_header.csv')
```

## 1. Missing Values Analysis
Checking for null or missing entries across all core datasets.

```python
datasets = {
    'Branches': branches_df,
    'Inventory Master': inventory_df,
    'Suppliers': suppliers_df,
    'Products': products_df,
    'Sales Orders': sales_df
}

for name, df in datasets.items():
    print(f'=== Missing Values: {name} ===')
    print(df.isnull().sum())
    print('\n')
```

## 2. Duplicate Record Check
Identifying redundant rows across primary keys and transaction tables.

```python
for name, df in datasets.items():
    duplicates = df.duplicated().sum()
    print(f'{name} duplicate row count: {duplicates}')
```

## 3. Data Types & Integrity Validation
Verifying schema types and checking negative numeric values.

```python
# Verify inventory and pricing non-negativity
print('Negative Stock Check:', (inventory_df['stock_level'] < 0).sum())
print('Negative Cost Check:', (products_df['unit_cost'] < 0).sum())
```
