```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams['font.sans-serif'] = 'DejaVu Sans'

print("=== STARTING WEEK 11: BUSINESS INTELLIGENCE DASHBOARD ===")

# 1. Consolidate Executive KPIs & Metrics
kpi_data = {
    'KPI_Metric': ['Total Revenue', 'Total Orders', 'Total Units Sold', 'Avg Order Value', 'Repeat Customer Rate', 'Forecasted Qtr Revenue'],
    'Value': [1842500, 12450, 48200, 147.99, 0.642, 520000],
    'Unit': ['USD', 'Count', 'Units', 'USD', 'Percentage', 'USD']
}
df_kpis = pd.DataFrame(kpi_data)
df_kpis.to_csv('../analysis/executive_kpi_summary.csv', index=False)
print("\n--- EXECUTIVE KPI CONSOLIDATION ---")
print(df_kpis.to_string(index=False))

# 2. Synthesize Integrated Business Data
np.random.seed(42)
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
historical_rev = np.random.normal(loc=150000, scale=15000, size=12).cumsum()
forecast_rev = historical_rev * 1.08

df_timeline = pd.DataFrame({'Month': months, 'Actual_Revenue': historical_rev, 'Forecast_Revenue': forecast_rev})

warehouses = ['Central Warehouse', 'East Hub', 'West Hub', 'North Hub']
wh_performance = pd.DataFrame({
    'Warehouse': warehouses,
    'Orders_Processed': [5600, 2800, 2400, 1650],
    'Fulfillment_Delay_Rate': [0.042, 0.019, 0.021, 0.015]
})

# 3. Render 6-Panel Executive BI Dashboard
fig, axes = plt.subplots(3, 2, figsize=(16, 14))

# Panel A: Actual vs Forecasted Revenue
axes[0, 0].plot(df_timeline['Month'], df_timeline['Actual_Revenue'], marker='o', label='Actual Revenue', color='navy')
axes[0, 0].plot(df_timeline['Month'], df_timeline['Forecast_Revenue'], marker='--', label='Forecast Target', color='teal')
axes[0, 0].set_title('Revenue Performance vs Forecast Trend')
axes[0, 0].legend()

# Panel B: Warehouse Fulfillment Volume
sns.barplot(data=wh_performance, x='Warehouse', y='Orders_Processed', ax=axes[0, 1], palette='Blues_r')
axes[0, 1].set_title('Orders Processed by Warehouse')

# Panel C: Warehouse Delay Rate Exception
sns.barplot(data=wh_performance, x='Warehouse', y='Fulfillment_Delay_Rate', ax=axes[1, 0], palette='Reds_r')
axes[1, 0].set_title('Warehouse Fulfillment Delay Rate (Exception Monitoring)')

# Panel D: Customer Concentration Deciles
deciles = [f'Decile {i}' for i in range(1, 11)]
rev_share = [0.58, 0.15, 0.09, 0.06, 0.04, 0.03, 0.02, 0.015, 0.01, 0.005]
axes[1, 1].bar(deciles, rev_share, color='darkgreen')
axes[1, 1].set_title('Customer Revenue Share by Decile')
axes[1, 1].tick_params(axis='x', rotation=45)

# Panel E: Inventory Reorder Status
categories = ['Cat A', 'Cat B', 'Cat C', 'Cat D']
current_stock = [1200, 3400, 2100, 800]
reorder_level = [1500, 2000, 1000, 1000]
x = np.arange(len(categories))
axes[2, 0].bar(x - 0.2, current_stock, 0.4, label='Current Stock', color='steelblue')
axes[2, 0].bar(x + 0.2, reorder_level, 0.4, label='Reorder Threshold', color='coral')
axes[2, 0].set_xticks(x)
axes[2, 0].set_xticklabels(categories)
axes[2, 0].set_title('Inventory Safety Stock Monitoring')
axes[2, 0].legend()

# Panel F: KPI Summary Table
axes[2, 1].axis('off')
table_data = df_kpis[['KPI_Metric', 'Value', 'Unit']].values
table = axes[2, 1].table(cellText=table_data, colLabels=['KPI Metric', 'Value', 'Unit'], loc='center', cellLoc='center')
table.scale(1, 1.8)
axes[2, 1].set_title('Executive KPI Scorecard Summary')

plt.tight_layout()
plt.show()
print("=== WEEK 11 BI DASHBOARD GENERATION COMPLETE ===")
```
