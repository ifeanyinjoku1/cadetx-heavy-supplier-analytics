```python
import pandas as pd
import numpy as np

# Load core relational datasets
branches_df = pd.read_csv('/content/branches.csv')
customers_df = pd.read_csv('/content/customers.csv')
inventory_df = pd.read_csv('/content/inventory_master.csv')
invoices_df = pd.read_csv('/content/invoices.csv')
payments_df = pd.read_csv('/content/payments.csv')
products_df = pd.read_csv('/content/products.csv')
purchase_orders_header_df = pd.read_csv('/content/purchase_orders_header.csv')
purchase_orders_lines_df = pd.read_csv('/content/purchase_orders_lines.csv')
sales_orders_header_df = pd.read_csv('/content/sales_orders_header.csv')
sales_orders_lines_df = pd.read_csv('/content/sales_orders_lines.csv')
stock_ledger_df = pd.read_csv('/content/stock_ledger.csv')
suppliers_df = pd.read_csv('/content/suppliers.csv')
```

```python
# Dictionary to manage all datasets efficiently
datasets = {
    'branches': branches_df,
    'customers': customers_df,
    'inventory': inventory_df,
    'invoices': invoices_df,
    'payments': payments_df,
    'products': products_df,
    'purchase_headers': purchase_orders_header_df,
    'purchase_lines': purchase_orders_lines_df,
    'sales_headers': sales_orders_header_df,
    'sales_lines': sales_orders_lines_df,
    'stock_ledger': stock_ledger_df,
    'suppliers': suppliers_df
}
```

```python
# ==========================================
# 1. Missing Values Analysis
# ==========================================
print("=== MISSING VALUES ANALYSIS ===")
for name, df in datasets.items():
    print(f"=== Missing Values: {name.upper()} ===")
    print(df.isnull().sum())
    print("\n")
```

```python
# ==========================================
# 2. Duplicate Record Check
# ==========================================
print("=== DUPLICATE RECORD CHECK ===")
for name, df in datasets.items():
    duplicates = df.duplicated().sum()
    print(f"{name.upper()} duplicate row count: {duplicates}")
print("\n")
```

```python
# ==========================================
# 3. Data Types & Integrity Validation
# ==========================================
print("=== DATA TYPES & INTEGRITY VALIDATION ===")

# Schema Types Summary
for name, df in datasets.items():
    print(f"--- Schema Types: {name.upper()} ---")
    print(df.dtypes)
    print("\n")

# Integrity Checks: Checking for negative values in critical numeric fields
print("--- Negative Value Checks ---")

# Check Inventory Quantities / Stock Levels
inv_qty_cols = [c for c in ['quantity', 'stock_level', 'qty_on_hand'] if c in inventory_df.columns]
if inv_qty_cols:
    col = inv_qty_cols[0]
    print(f"Negative Stock Check ({col}):", (inventory_df[col] < 0).sum())

# Check Product Costs / Prices
prod_cost_cols = [c for c in ['unit_cost', 'cost', 'unit_price', 'price'] if c in products_df.columns]
if prod_cost_cols:
    col = prod_cost_cols[0]
    print(f"Negative Cost/Price Check ({col}):", (products_df[col] < 0).sum())

# Check Sales Lines Quantities & Unit Prices
if 'quantity' in sales_orders_lines_df.columns:
    print("Negative Sales Quantity Check:", (sales_orders_lines_df['quantity'] < 0).sum())

if 'unit_price' in sales_orders_lines_df.columns:
    print("Negative Sales Price Check:", (sales_orders_lines_df['unit_price'] < 0).sum())
```
